```python
import matplotlib.pyplot as plt
import numpy as np


rows = 15
cols = 15

for idx, d in enumerate(np.linspace(0, 1, 60)):

    fig = plt.figure(figsize=(8, 8))

    plot_size = 1.0 / max(rows, cols)

    for i in range(rows):
        for j in range(cols):

            x_pos = j * plot_size
            y_pos = 1 - (i + 1) * plot_size

            ax = fig.add_axes([x_pos, y_pos, plot_size, plot_size])

            wave_position = (i + j) / (rows + cols)
            fraction = (np.sin(2 * np.pi * (wave_position - d)) + 1) / 2

            fraction = max(0, min(1, fraction))
            sizes = [fraction, 1 - fraction]
            
            # Создаем pie chart
            ax.pie(sizes, 
                colors=['black', 'lightgray'],
                startangle=90,
                wedgeprops={'edgecolor': 'none', 'linewidth': 0})
            
            ax.axis('off')

    plt.savefig(f'output/image_{idx}.png', bbox_inches='tight', pad_inches=0, dpi=150)
    plt.clf()
    plt.close()

    print(idx, end=', ')
```

```python
import imageio.v2 as imageio
import glob
import re


image_files = glob.glob('output/image_*.png')
image_files.sort(key=lambda x: int(re.search(r'image_(\d+)\.png', x).group(1)))

print(f"Найдено {len(image_files)} файлов")

images = []
for filename in image_files:
    images.append(imageio.imread(filename))

imageio.mimsave('animation.gif', images, fps=15, loop=0)
```
