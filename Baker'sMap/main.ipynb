```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 2000
arr = np.zeros((N, N))
```

```python
for i in range(N):
    arr[i] = np.array([i + j for j in range(N)])
```

```python
plt.figure(figsize=(16, 8))
plt.imshow(arr, cmap='hot', origin='lower')
plt.colorbar()
plt.plot()
```

```python
def baker(x, y): 
    if 0 <= x < 0.5: 
        return 2 * x, y / 2
    else:
        return 2 - 2 * x, 1 - y / 2
```

```python
for _ in range(8):

    new_arr = np.zeros_like(arr)

    for i in range(N): 
        for j in range(N): 
            x, y = baker(i / N, j / N)
            new_arr[i, j] = arr[int(x * N) - 1, int(y * N) - 1]

    arr = new_arr.copy()
```

```python
plt.figure(figsize=(16, 8))
plt.imshow(arr, cmap='hot', origin='lower')
plt.colorbar()
plt.plot()
```
