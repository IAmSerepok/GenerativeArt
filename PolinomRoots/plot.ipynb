```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 14

# Список цветов для каждого корня
colors = plt.cm.rainbow(np.linspace(0, 1, N))

# Визуализация с разными цветами для каждого корня
plt.figure(figsize=(16, 9))

alpha = 0.03
s = 0.1

# Загружаем данные с индексами
try:
    data = np.loadtxt('data.csv', delimiter=',')
    if len(data) > 0:
        real_parts = data[:, 0]
        imag_parts = data[:, 1]
        indices = data[:, 2].astype(int)
        
        # Рисуем каждый корень своим цветом
        for root_idx in range(N):
            mask = indices == root_idx
            if np.any(mask):
                plt.scatter(real_parts[mask], imag_parts[mask], 
                           color=colors[root_idx], 
                           edgecolors='none', 
                           alpha=alpha, s=s)
                
except MemoryError:
    # Построчная обработка для экономии памяти
    from collections import defaultdict
    
    # Словарь для хранения данных по индексам
    root_data = defaultdict(list)
    chunk_size = 100_000
    
    with open('data.csv', 'r') as f:
        chunk = []
        for line in f:
            parts = line.strip().split(',')
            if len(parts) == 3:
                real_val = float(parts[0])
                imag_val = float(parts[1])
                idx = int(float(parts[2]))
                root_data[idx].append((real_val, imag_val))
        
        # Рисуем для каждого индекса
        for root_idx in range(N):
            if root_idx in root_data and root_data[root_idx]:
                data = np.array(root_data[root_idx])
                plt.scatter(data[:, 0], data[:, 1], 
                           color=colors[root_idx], 
                           edgecolors='none', 
                           alpha=alpha, s=s)

plt.xlim(-3.5, 3.5)
plt.ylim(-2.5, 2)

plt.axis('off')

plt.tight_layout()
plt.savefig('output/polynomial_roots.png', dpi=200, bbox_inches='tight')
plt.show()
```
