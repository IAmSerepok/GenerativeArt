```python
import cv2
import os

# Папка с кадрами
frames_folder = 'output'
# Сортируем кадры по числу в названии
frames_list = sorted(
    [f for f in os.listdir(frames_folder) if f.endswith(('.png', '.jpg', '.jpeg'))],
    key=lambda f: int(''.join(filter(str.isdigit, f)))
)

# Получаем размер первого кадра
first_frame = cv2.imread(os.path.join(frames_folder, frames_list[0]))
height, width, _ = first_frame.shape  # (height, width, channels)

# Создаём видео (MP4 с кодеком H.264)
output_file = 'output.mp4'
fourcc = cv2.VideoWriter_fourcc(*'H264')  # или 'avc1' (H.264), но может требовать доп. настройки
video = cv2.VideoWriter(output_file, fourcc, 30.0, (width, height))

# Записываем кадры
for frame_name in frames_list:
    frame_path = os.path.join(frames_folder, frame_name)
    frame = cv2.imread(frame_path)
    video.write(frame)

video.release()
print(f"Видео сохранено как {output_file}")
```
