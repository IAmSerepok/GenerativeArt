```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import CubicSpline
```

```python
plt.figure(figsize=(8, 4))

t = np.linspace(0, 1, 2)

for y0 in range(10):
    plt.plot(t, len(t) * [y0], color='white')

plt.axis('off')
plt.tight_layout()
plt.savefig('assets/lines/img_1.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 4))

t = np.linspace(0, 1, 2)

for y0 in range(10):
    y = np.random.uniform(-0.4, 0.4) * t + y0
    plt.plot(t, y, color='white')

plt.axis('off')
plt.tight_layout()
plt.savefig('assets/lines/img_2.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 4))

t = np.linspace(0, 1, 20)

for y0 in range(10):
    noise = np.random.uniform(-0.25, 0.25, len(t))

    y = np.random.uniform(-0.4, 0.4) * t + y0 + noise
    
    plt.plot(t, y, color='white')

plt.axis('off')
plt.tight_layout()
plt.savefig('assets/lines/img_3.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 4))

t = np.linspace(0, 1, 20)

for y0 in range(10):
    noise = np.random.uniform(-0.25, 0.25, len(t))

    y = np.random.uniform(-0.4, 0.4) * t + y0 + noise

    cs = CubicSpline(t, y, bc_type='natural')

    new_t = np.linspace(0, 1, 100)
    
    plt.plot(new_t, cs(new_t), color='white')

plt.axis('off')
plt.tight_layout()
plt.savefig('assets/lines/img_4.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 4))

t = np.linspace(0, 1, 20)

for y0 in range(10):
    noise = np.random.uniform(-0.25, 0.25, len(t))

    y = np.random.uniform(-0.4, 0.4) * t + y0 + noise

    cs = CubicSpline(t, y, bc_type='natural')

    new_t = np.linspace(0, 1, 100)
    new_y = cs(new_t)

    for _ in range(10):
        smooth_y = new_y.copy()
        for idx in range(1, len(new_t) - 1):
            smooth_y[idx] = new_y[idx - 1 : idx + 2].sum() / 3
        new_y = smooth_y.copy()
    
    plt.plot(new_t, new_y, color='white')

plt.axis('off')
plt.tight_layout()
plt.savefig('assets/lines/img_5.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 4))

t = np.linspace(0, 1, 20)

for y0 in range(3):
    y = np.random.uniform(-0.3, 0.3) * t + y0 + np.random.uniform(-0.05, 0.05, len(t))

    # Основная линия
    cs = CubicSpline(t, y, bc_type='natural')

    new_t = np.linspace(0, 1, 500)
    new_y = cs(new_t)

    plt.plot(new_t, new_y, color='white', linewidth=1)

    # Текстура (Песочный сплайн)
    for _ in range(50):
        noise = np.random.uniform(-0.02, 0.02, len(t))

        cs = CubicSpline(t, y + noise, bc_type='natural')

        new_t = np.linspace(0, 1, np.random.randint(500, 2000))
        new_y = cs(new_t)
        
        plt.scatter(new_t, new_y, color='white', s=0.05, alpha=0.9, edgecolor='none')

plt.axis('off')
plt.tight_layout()
plt.savefig('assets/lines/img_6.png', dpi=300)
plt.show()
```
