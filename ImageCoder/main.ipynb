```python
import matplotlib.pyplot as plt
import numpy as np

from PIL import Image
```

```python
def halton_sequence(b):
    """Generator function for Halton sequence."""
    n, d = 0, 1
    while True:
        x = d - n
        if x == 1:
            n = 1
            d *= b
        else:
            y = d // b
            while x <= y:
                y //= b
            n = (b + 1) * y - x
        yield n / d
```

```python
size = (1000, 1000, 3)
```

```python
# field = np.zeros(size, dtype=np.uint8)

field = np.random.randint(0, 255, size, dtype=np.uint8)
```

```python
images = [
    [17, 31, 'input/test.png'], 
    [29, 71, 'input/test2.png'], 
]
```

```python
for index, (seed_x, seed_y, path) in enumerate(images):
    halton_x = halton_sequence(seed_x)
    halton_y = halton_sequence(seed_y)

    img = Image.open(path)
    img = img.convert('RGB')
    pixels = np.array(img).astype(np.uint8)
    print(pixels.shape)

    for i in range(pixels.shape[0]):
        for j in range(pixels.shape[1]): 
            x = int(next(halton_x) * size[0])
            y = int(next(halton_y) * size[1])

            field[x, y, ::] = pixels[i, j, ::]
```

```python
shapes = [
    (1024, 683, 3), 
    (456, 811, 3)
]
seeds = [
    (17, 31), (29, 71)
]
```

```python
img = Image.open('coded_image.png')
img = img.convert('RGB')
pixels = np.array(img).astype(np.uint8)
size = pixels.shape
```

```python
for idx, (shape, seed) in enumerate(zip(shapes, seeds)):
    decode_pixels = np.zeros(shape, dtype=np.uint8)

    halton_x = halton_sequence(seed[0])
    halton_y = halton_sequence(seed[1])

    for i in range(decode_pixels.shape[0]):
        for j in range(decode_pixels.shape[1]): 

            x = int(next(halton_x) * size[0])
            y = int(next(halton_y) * size[1])

            decode_pixels[i, j, ::] = pixels[x, y, ::]

    Image.fromarray(decode_pixels, 'RGB').save(f'output/decoded_image_{idx}.png')
```
